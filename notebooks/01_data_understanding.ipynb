```python
# Missing-value assessment

missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0].sort_values(ascending=False)

print("Columns with missing values:")
print(missing_values)
```

```python
# Missing-value percentage

missing_percentage = (
    df.isnull().mean() * 100
).sort_values(ascending=False)

missing_percentage = missing_percentage[missing_percentage > 0]

print("Missing-value percentage:")
print(missing_percentage)
```

### Numerical Variable Findings

The numerical investigation identified 29 numerical columns. These include identifiers, business measures, operational metrics, and geographic variables.

Several important observations were identified:

- Identifier columns such as `Order Id`, `Customer Id`, `Product Card Id`, and `Order Item Id` are numeric but represent entities rather than continuous measurements.
- `Order Item Quantity` ranges from 1 to 5, with a median of 1.
- `Product Price` and `Order Item Product Price` range from approximately 9.99 to 1999.99.
- `Late_delivery_risk` is a binary variable containing values from 0 to 1.
- `Days for shipping (real)` ranges from 0 to 6 days.
- `Days for shipment (scheduled)` ranges from 0 to 4 days.
- Profit-related measures contain both positive and negative values. Negative profit observations will not be treated as errors without further investigation because they may represent genuinely unprofitable transactions.
- `Order Zipcode` has a high proportion of missing values and will require careful treatment.
- `Product Description` contains no non-null observations and is therefore a strong candidate for removal during the cleaning stage.
- No conclusions about numerical outliers or invalid business values will be made solely from descriptive statistics. Potential anomalies will be investigated using business rules and cross-column relationships before cleaning.

The numerical investigation therefore identifies several areas requiring validation rather than immediate correction.

```python
# Validate relationships between important numerical fields

print("Sales - Discount = Order Item Total:")

total_difference = (
    df["Sales"] - df["Order Item Discount"] - df["Order Item Total"]
).abs()

print("Maximum absolute difference:", total_difference.max())
print("Rows with difference > 0.01:", (total_difference > 0.01).sum())


print("\nProduct Price vs Order Item Product Price:")

price_difference = (
    df["Product Price"] - df["Order Item Product Price"]
).abs()

print("Maximum absolute difference:", price_difference.max())
print("Rows with difference > 0.01:", (price_difference > 0.01).sum())
```

```python
# Validate quantity and delivery-risk ranges

print("Unique Order Item Quantity values:")
print(sorted(df["Order Item Quantity"].unique()))

print("\nUnique Late Delivery Risk values:")
print(sorted(df["Late_delivery_risk"].unique()))

print("\nUnique Product Status values:")
print(sorted(df["Product Status"].unique()))
```

```python
# Check whether Sales is consistent with price × quantity

sales_difference = (
    df["Order Item Product Price"] * df["Order Item Quantity"]
    - df["Sales"]
).abs()

print("Maximum absolute difference between price × quantity and Sales:",
      sales_difference.max())

print("Rows with difference > 0.01:",
      (sales_difference > 0.01).sum())
```

```python
# Validate Sales - Discount = Order Item Total after rounding

total_difference_rounded = (
    (
        df["Sales"]
        - df["Order Item Discount"]
        - df["Order Item Total"]
    )
    .round(2)
    .abs()
)

print("Maximum difference after rounding:",
      total_difference_rounded.max())

print("Rows with difference > 0.01 after rounding:",
      (total_difference_rounded > 0.01).sum())
```
